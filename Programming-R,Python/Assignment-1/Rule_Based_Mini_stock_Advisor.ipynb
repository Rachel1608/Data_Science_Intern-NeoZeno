### Assignment 1
Implementation of Rule Based Mini Stock Advisor, to generate Buy/Sell/Hold recommendations for the given Stock Data.

```python
import pandas as pd
```

```python
# Dataset 

data = [ 
    [1, 101, 104],
    [2, 103, 101],
    [3, 115, 114],
    [4, 105, 107],
    [5, 101, 105],
    [6, 89, 76],
    [7, 91, 98],
    [8, 200, 250],
    [9, 190, 200],
    [10, 320, 260]    
]

columns = ['Days','opening_price', 'closing_price']

df = pd.DataFrame(data, columns = columns)

df
```

```python
# Compute the daily percent change for each record in a dataset of 10 days. 
for index, row in df.iterrows():
    opening_price = row['opening_price']
    closing_price = row['closing_price']
    percentage_change = ((closing_price - opening_price) / opening_price) * 100
    df.at[index, 'percentage_change'] = percentage_change
    print(f"Day {index}: {percentage_change:.2f}%")
```

```python
#Apply the rule-based system to classify each day as BUY, SELL, or HOLD. 
for i in range(len(df)):
    percent = df.iloc[i]['percentage_change']
    prev_percent = df.iloc[i-1]['percentage_change'] if i > 0 else 0  # Yesterday's change (0 for day 1)
    
    # BUY: >2% OR (positive today AND positive yesterday)
    if percent > 2 or (percent > 0 and prev_percent > 0):
        classification = 'BUY'
        
    # SELL: <-2% OR (negative today AND negative yesterday)  
    elif percent < -2 or (percent < 0 and prev_percent < 0):
        classification = 'SELL'
    else:
        classification = 'HOLD'
    
    # Storing the classification in the DataFrame 
    df.at[i, 'classification'] = classification
    print(f"Day {i}: {classification}")
```
