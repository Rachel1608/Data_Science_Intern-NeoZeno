```python
# Let's do the prediction based on Rule-based Condition
def predict_exam_performance(row):
    if row['course'].lower() == 'bca':
        if row['exam_score'] >= 85:
            return "bca Top Scorer"
        elif row['exam_score'] < 85 and row['exam_score'] >= 50:
            return "bca Average Scorer"
        else:
            return "bca Poor Scorer"
    elif row['course'].lower() == 'b.tech':
        if row['exam_score'] >= 85:
            return "b.tech Top Scorer"
        elif row['exam_score'] < 85 and row['exam_score'] >= 50:
            return "b.tech Average Scorer"
        else:
            return "b.tech Poor Scorer"
    elif row['course'].lower() == 'b.com':
        if row['exam_score'] >= 85:
            return "b.com Top Scorer"
        elif row['exam_score'] < 85 and row['exam_score'] >= 50:
            return "b.com Average Scorer"
        else:
            return "b.com Poor Scorer"
    elif row['course'].lower() == 'bba':
        if row['exam_score'] >= 85:
            return "bba Top Scorer"
        elif row['exam_score'] < 85 and row['exam_score'] >= 50:
            return "bba Average Scorer"
        else:
            return "bba Poor Scorer"
    elif row['course'].lower() == 'diploma':
        if row['exam_score'] >= 85:
            return "diploma Top Scorer"
        elif row['exam_score'] < 85 and row['exam_score'] >= 50:
            return "diploma Average Scorer"
        else:
            return "diploma Poor Scorer"
    elif row['course'].lower() == 'b.sc':
        if row['exam_score'] >= 85:
            return "b.sc Top Scorer"
        elif row['exam_score'] < 85 and row['exam_score'] >= 50:
            return "b.sc Average Scorer"
        else:
            return "b.sc Poor Scorer"
    elif row['course'].lower() == 'ba':
        if row['exam_score'] >= 85:
            return "ba Top Scorer"
        elif row['exam_score'] < 85 and row['exam_score'] >= 50:
            return "ba Average Scorer"
        else:
            return "ba Poor Scorer"
    else:
        return "Other Course"
```

```python
# Create new column prediction to apply the defined function into the dataset
data['Prediction'] = data.apply(predict_exam_performance, axis = 1)
```
