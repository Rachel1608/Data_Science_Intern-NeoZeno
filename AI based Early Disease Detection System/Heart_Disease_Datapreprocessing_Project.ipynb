# Heart Disease Dataset
This data set dates from 1988 and consists of four databases: Cleveland, Hungary, Switzerland, and Long Beach V. It contains 76 attributes, including the predicted attribute, but all published experiments refer to using a subset of 14 of them. The "target" field refers to the presence of heart disease in the patient. It is integer valued 0 = no disease and 1 = disease.

**Attribute Information:**
* age - age
* sex - sex
* cp - chest pain type (4 values)
* trestbps - resting blood pressure
* chol - serum cholestoral in mg/dl
* fbs - fasting blood sugar > 120 mg/dl
* restecg - resting electrocardiographic results (values 0,1,2)
* thalach - maximum heart rate achieved
* exang - exercise induced angina
* oldpeak - ST depression induced by exercise relative to rest
* slope - the slope of the peak exercise ST segment
* ca - number of major vessels (0-3) colored by flourosopy
* thal -  0 = normal; 1 = fixed defect; 2 = reversable defect
* target

# Data Preprocessing

```python
# Lets check the skewness
data.skew()
```

* From the Continuous variables skewness ***resting blood pressure(trestbps)*** is **moderate right skew.**
* ***Cholestrol (chol)*** is **right skewed**
* ***maximum heart rate achieved(thalach)*** is **Mild left skew**
* ***oldpeak*** is **high right skew**


```python
#Checking the Skewness 
plt.figure(figsize=(20,15))
plotnumber=1
for column in data:
    if plotnumber<=14:
        ax=plt.subplot(6,3,plotnumber)
        sns.distplot(data[column])
        plt.xlabel(column,fontsize=20)
    plotnumber +=1
plt.tight_layout()
```

```python
# Check the outliers
plt.figure(figsize=(20,15))
plotnumber=1
for column in data:
    if plotnumber<=14:
        ax=plt.subplot(6,3,plotnumber)
        sns.boxplot(data[column])
        plt.xlabel(column,fontsize=20)
    plotnumber +=1
plt.tight_layout()
```

* From visualization we clearly understand trestbps, chol, thalach, oldpeak needs to be treated.
* We skip thalach since it is mild left skew.

```python
# Treat the outliers and skewness
#1st Quantile
q1 = data.quantile(0.25)
```

```python
#3rd Quantile
q3 = data.quantile(0.75)
```

```python
# IQR Range
IQR = q3 - q1
```

```python
print("1st Quantile:")
print(q1)
print(" ")
print("3rd Quantile:")
print(q3)
print(" ")
print("Inter Quartile Range:")
print(IQR)
```

```python
### Outliers Detection Formula

#### Upper Limit: Q3 +(1.5 * IQR)

#### Lower Limit: Q1 -(1.5 * IQR)
```

```python
#Validating Resting Blood pressure outlier
trest_pred_high = (q3.trestbps + (1.5 * IQR.trestbps))
trest_pred_high
```

```python
#Check the indexes which have higher values
index=np.where(data['trestbps']>trest_pred_high)
index
```

```python
data['trestbps'] = np.sqrt(data['trestbps'])
```

```python
sns.distplot(data['trestbps'])
```

```python
sns.boxplot(data['trestbps'])
```

```python
#Validating cholestrol outlier
chol_pred_high = (q3.chol + (1.5 * IQR.chol))
chol_pred_high
```

```python
#Check the indexes which have higher values
index=np.where(data['chol']>chol_pred_high)
index
```

```python
data['chol'] = np.log1p(data['chol'])
```

```python
sns.distplot(data['chol'])
```

```python
sns.boxplot(data['chol'])
```

```python
#Validating oldpeak outlier
op_pred_high = (q3.oldpeak + (1.5 * IQR.oldpeak))
op_pred_high
```

```python
#Check the indexes which have higher values
index=np.where(data['oldpeak']>op_pred_high)
index
```

```python
data['oldpeak'] = np.log1p(data['oldpeak'])
```

```python
sns.distplot(data["oldpeak"])
```

```python
# Checking the Skewness after treating outliers
plt.figure(figsize=(20,15))
plotnumber=1
for column in data:
    if plotnumber<=14:
        ax=plt.subplot(6,3,plotnumber)
        sns.distplot(data[column])
        plt.xlabel(column,fontsize=20)
    plotnumber +=1
plt.tight_layout()
```

```python
# Checking the Skewness after treating outliers
plt.figure(figsize=(20,15))
plotnumber=1
for column in data:
    if plotnumber<=14:
        ax=plt.subplot(6,3,plotnumber)
        sns.boxplot(data[column])
        plt.xlabel(column,fontsize=20)
    plotnumber +=1
plt.tight_layout()
```

* From the distplot and boxplot, we clearly understand the outliers and skewness are treated well.
* For ***trestbps*** we used **square root transformation**.
* ***cholestrol and oldpeak*** we used **Log Transformation***.
*  IQR to treat the outliers.
