## Data Preprocessing

```python
# Symptom columns
symptom_cols = [f'Symptom_{i}' for i in range(1, 18)]

# Convert symptom columns into list
data['symptom_list'] = data[symptom_cols].values.tolist()

# Remove NaN values
data['symptom_list'] = data['symptom_list'].apply(
    lambda x: [symptom for symptom in x if pd.notna(symptom)]
)
```

```python
# Encode symptoms
mlb = MultiLabelBinarizer()

symptom_encoded = pd.DataFrame(
    mlb.fit_transform(data['symptom_list']),
    columns=mlb.classes_
)
```

```python
# Encode target variable (Disease)
le = LabelEncoder()

symptom_encoded['Disease'] = le.fit_transform(data['Disease'])
```

```python
# Save final encoded dataset
symptom_encoded.to_csv("encoded_disease_dataset.csv", index=False)

symptom_encoded.head()
```
