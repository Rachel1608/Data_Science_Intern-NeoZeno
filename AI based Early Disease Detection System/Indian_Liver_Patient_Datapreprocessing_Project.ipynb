## Data Preprocessing
* As already mentioned above we observed skewness in the dataset, let's transform the skewness and ouliers using Log Transformation method and IQR Inter Quartile Range.

```python
# Checking the Skewness 
plt.figure(figsize=(20,15))
plotnumber=1
for column in data:
    if plotnumber<=11:
        ax=plt.subplot(4,3,plotnumber)
        sns.distplot(data[column])
        plt.xlabel(column,fontsize=20)
    plotnumber +=1
plt.tight_layout()
```

***The distplot Visualization clearly shows that there is Highly Right skewed in the "Total_Bilirubin, Direct Bilirubin, Alkaline_Phosphotase, Alamine_Aminotransferase, Aspartate_Aminotransferase"***.

```python
# Check the outliers
plt.figure(figsize=(20,15))
plotnumber=1
for column in data:
    if plotnumber<=11:
        ax=plt.subplot(4,3,plotnumber)
        sns.boxplot(data[column])
        plt.xlabel(column,fontsize=20)
    plotnumber +=1
plt.tight_layout()
```

***The boxplot Visualization clearly shows that there is outliers present in the "Total_Bilirubin, Direct Bilirubin, Alkaline_Phosphotase, Alamine_Aminotransferase, Aspartate_Aminotransferase" which needs to be treated***.

```python
data.skew()
```

* Apart from Binary values, we'll understand the skewness from Continous Variables
* **Total_Bilirubin, Direct_Bilirubin, Alkaline_Phosphotase, Alamine_Aminotransferase, Aspartate_Aminotransferase** are highly Right skewed Attributes.
* **Albumin_and_Globulin_Ratio** is moderate skewed.

* There are no zero values let's move with treating outliers and skewness.

```python
# Treat the Outliers
# Let's find the IQR (Inter-Quartile Range) to identify the outliers
```

```python
#1st Quantile
q1 = data.quantile(0.25)
```

```python
#3rd Quantile
q3 = data.quantile(0.75)
```

```python
# IQR Range
IQR = q3 - q1
```

### *Outliers Detection Formula**

#### ***Higher side:*** Q3 +(1.5 * IQR)


#### ***Lower Side:*** Q1 -(1.5 * IQR)

```python
print("1st Quantile:")
print(q1)
print(" ")
print("3rd Quantile:")
print(q3)
print(" ")
print("Inter Quartile Range:")
print(IQR)
```

```python
#Validating Total Bilirubin Outlier
bil_pred_high = (q3.Total_Bilirubin + (1.5 * IQR.Total_Bilirubin))
bil_pred_high
```

```python
# Index values of Total_Bilirubin Higher Values
index=np.where(data['Total_Bilirubin']>bil_pred_high)
index
```

```python
# Log Transformation Method
data['Total_Bilirubin'] = np.log1p(data['Total_Bilirubin'])
```

```python
data['Total_Bilirubin'] = np.where(data['Total_Bilirubin'] > bil_pred_high, bil_pred_high, data['Total_Bilirubin'])
```

```python
data.head()
```

```python
sns.boxplot(data["Total_Bilirubin"])
```

## Multicollinearity Problem

If there are 2 independent features are highly correlated with each other is known as Multicollinearity Problem, which leads to overfitting problem and affects the model stability. There are several ways to check Multicollinearity such as using Heatmap, Variance Inflation Factor(VIF). We'll check the Multicollinearity problem using heatmap.

```python
data_corr = data.corr()
```

```python
plt.figure(figsize=(10,8))

sns.heatmap(
    data_corr,
    annot=True,
)

plt.title("Correlation Heatmap")
plt.show()
```

* We can clearly see the Heatmap visulation Total_Bilirubin and Direct_Bilirubin are highly correlated with each other around **98%** this affects the model's performance.
* So we either drop any one of the feature either Total_Bilirubin or Direct_Bilirubin.
* Let's drop Direct_Bilirubin.

```python
df = data.drop('Direct_Bilirubin', axis=1)
```
