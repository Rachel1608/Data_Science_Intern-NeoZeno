```python
import nltk
from nltk.tokenize import  word_tokenize
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer, WordNetLemmatizer
from sklearn.feature_extraction.text import TfidfVectorizer
from gensim.models import Word2Vec
```

```python
nltk.download('punkt')
nltk.download('stopwords')
nltk.download('wordnet')
```

```python
#Stopwords

stopwords = set(stopwords.words("english"))
stemmer = PorterStemmer()
lemmatizer = WordNetLemmatizer()
```

```python
clean_reviews = []
for text in data['review']:
    words = word_tokenize(text.lower())
    words = [w for w in words if w not in stopwords]
    words = [stemmer.stem(w) for w in words]
    words = [lemmatizer.lemmatize(w) for w in words]
    clean_reviews.append("".join(words))
print(clean_reviews)
```

```python
#Vectorizer
vectorizer = TfidfVectorizer(ngram_range = (1,2))
x = vectorizer.fit_transform(clean_reviews)
print(x.toarray())
```

```python
tokens = [review.split() for review in clean_reviews]
model = Word2Vec(tokens, vector_size = 50, window = 2, min_count = 1)
print("words in vocabulary:", model.wv.index_to_key)
print("vector of first words:", model.wv[model.wv.index_to_key[0]])
```
